# **What is Autograd?**

Autograd is a core component of PyTorch that provides automatic differentiation for tensor operations. It enables gradient computation, which is essential for training machine learning models using optimization algorithms like gradient descent.

```python
import torch
```

```python
x = torch.tensor(3.0,requires_grad = True)
# requires grad is by default false it basically lets pytorch know that in future we want the gradient of this tensor
```

```python
y=x**2
```

```python
x
```

```python
y
# you can see in the output "grad_fn=<PowBackward0>)" this basically tells that when we go backwards it will calculate the
#  derivative , it is due to the requires_grad function we used earlier
```

```python
# calculating grad
y.backward()
```

```python
x.grad
# you can see the output as 6 like the derivative of x**2 is 2x so if x=3 then 2x = 6
```

### Clearing Gradients — Revision Notes
During backpropagation, PyTorch stores gradients in **".grad."**
If we call **".backward()"** multiple times, gradients accumulate instead of being replaced.
Therefore, before the next backpropagation, we usually clear/reset the gradients

### **Example 1: Gradient accumulation**
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)    **Output:  tensor(4.)**


### **Now call backward again:**

y = x ** 2

y.backward()

print(x.grad)    **Output :  tensor(8.)**
