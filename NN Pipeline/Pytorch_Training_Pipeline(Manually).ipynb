```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
```

```python
df.drop(columns=["id","Unnamed: 32"],inplace=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0:],test_size=0.2)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
encoder = LabelEncoder()
y_train=encoder.fit_transform(y_train.iloc[:,0])
y_test=encoder.transform(y_test.iloc[:,0])
```

## **Numpy Array To Pytorch Tensors**

```python
X_train_tensors = torch.from_numpy(X_train)
X_test_tensors = torch.from_numpy(X_test)
y_train_tensors = torch.from_numpy(y_train)
y_test_tensors = torch.from_numpy(y_test)
```

```python
X_train_tensors.shape
```

```python
y_train_tensors.shape
```

```python
class MySimpleNN():
  def __init__(self,X):
    self.weights = torch.rand(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias = torch.zeros(1,dtype=torch.float64,requires_grad=True)

  def forward(self,X):
    return torch.sigmoid(torch.matmul(X,self.weights)+self.bias)

  def loss_function(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
learning_rate = 0.1
epochs=25
```

```python
model = MySimpleNN(X_train_tensors)

for epoch in range(epochs):
  # Forward Pass
  y_pred = model.forward(X_train_tensors)

  # Loss Calculate
  loss = model.loss_function(y_pred,y_train_tensors)

  # Backward Pass
  loss.backward()

  # Update Weights
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # Zero Gradients
    model.weights.grad.zero_()
    model.bias.grad.zero_()

  # Print Loss In Each Epoch
  print(f"Epoch: {epoch+1}, Loss: {loss.item()}")
```

```python
# model evaluation
with torch.no_grad():
    y_pred = model.forward(X_test_tensors)
    y_pred = (y_pred > 0.73).float()
    accuracy = (y_pred == y_test_tensors).float().mean()
    print(f'Accuracy: {accuracy.item()}')
```
